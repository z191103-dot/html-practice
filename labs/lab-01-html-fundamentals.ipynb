## Setup — Run this cell first
This installs the HTML parser and creates reusable preview/testing helpers.

```python
%pip -q install beautifulsoup4
from bs4 import BeautifulSoup
from IPython.display import HTML, display
from html import escape

def preview(source):
    # iframe isolates the rendered student page from Colab's interface
    display(HTML('<iframe sandbox="allow-scripts" style="width:100%;height:260px;border:1px solid #aaa;border-radius:8px" srcdoc="' +
                 escape(source, quote=True) + '"></iframe>'))

def check(source, tests):
    soup = BeautifulSoup(source, "html.parser")
    passed = 0
    for label, test in tests:
        try:
            ok = bool(test(soup))
        except Exception:
            ok = False
        print(("✅ PASS" if ok else "❌ FIX ") + " — " + label)
        passed += ok
    print(f"\nResult: {passed}/{len(tests)} checks passed")
    return passed == len(tests)

print("Setup complete! Continue to Exercise 1.")
```

## Exercise 1 — Your First HTML Document

HTML provides the **structure** of a webpage. A complete document typically has `<!DOCTYPE html>`, `<html>`, `<head>`, and `<body>`.

**Your task**
- Set the document language to English.
- Add a page title inside `<head>`.
- Add one `<h1>` and one `<p>` inside `<body>`.

**Tip:** Change the placeholder comments below; don't delete the outer document structure.

```python
html_1 = '''
<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <title>My First Page</title>
</head>
<body>
  <h1>Hello world</h1>
  <p>Hey</p>
</body>
</html>
'''
```

```python
preview(html_1)
check(html_1, [
    ("HTML document has a doctype", lambda s: html_1.lstrip().lower().startswith("<!doctype html>")),
    ("HTML language is English", lambda s: s.html and s.html.get("lang") == "en"),
    ("Page includes a non-empty title", lambda s: s.title and s.title.get_text(strip=True)),
    ("Body contains an h1 with text", lambda s: s.body and s.body.h1 and s.body.h1.get_text(strip=True)),
    ("Body contains a paragraph with text", lambda s: s.body and s.body.p and s.body.p.get_text(strip=True)),
])
```

## Exercise 2 — Text and Lists

Headings communicate hierarchy. Lists group related items.

**Your task**
- Add an `<h2>` titled **My Hobbies**.
- Add an unordered list `<ul>` with at least **three** `<li>` items.
- Add a paragraph containing `<strong>` or `<em>` text.

```python
html_2 = '''
<section>
  <h2>My hobbies</h2>
  <ul>
  <li/>Watch footbal</li>
  <li>Play tennis</li>
  <li>I enjoy learning about <strong>AI</strong></li>
  <p>This is my hobbies</p>
  </ul>
</section>
'''
```

```python
preview(html_2)
check(html_2, [
    ("An h2 says My Hobbies", lambda s: any(h.get_text(" ", strip=True).lower() == "my hobbies" for h in s.find_all("h2"))),
    ("An unordered list exists", lambda s: s.ul is not None),
    ("The list has at least 3 items", lambda s: s.ul is not None and len(s.ul.find_all("li", recursive=False)) >= 3),
    ("A paragraph includes emphasized or strong text", lambda s: any(p.find(["strong", "em"]) for p in s.find_all("p"))),
])
```
